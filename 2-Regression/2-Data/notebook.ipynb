```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets, linear_model, model_selection

#Load the database diabetes

X, y = datasets.load_diabetes(return_X_y=True)

#Print the shape of the data and the first row.

print(X)
# print(y)
print(X[0])
```

```python
#Extract the column at index 2

X = X[:, 2]
print(X.shape)

#Reshape to a 2D array.

X = X.reshape(-1, 1)
print(X.shape) 
```

```python
# Split the data into training and testing data

X_train, X_test, y_train, y_test = model_selection.train_test_split(X, y, test_size=0.33)
```

```python
# Create a linear regression model and train it with our data.
model = linear_model.LinearRegression()
model.fit(X_train, y_train)
```

```python
# Predict our testing data

y_pred = model.predict(X_test)
```

```python
# Create a scatter plot 
plt.scatter(X_test, y_test, color='black')

# Plot the predictions
plt.plot(X_test, y_pred, color='blue', linewidth=3)

# Add labels and a title
plt.xlabel('Scaled BMIs')
plt.ylabel('Disease Progression')
plt.title("A Graph Plot showing Diabetes Progression against BMIs")

# Draw the plot
plt.show()
```

```python
import pandas as pd

# Read the US pumpkins CVS file

pumpkins = pd.read_csv('../data/US-pumpkins.csv')

# Filter out any rows that don't use price per bushel.

pumpkins = pumpkins[pumpkins['Package'].str.contains('bushel', case=True, regex=True)]
pumpkins.head()
```

```python
#Count how many cell in each column are empty.

pumpkins.isnull().sum()
```

```python
# Define the columns we want to keep data

new_columns = ['Package', 'Month', 'Low Price', 'High Price', 'Date']

# Drop all the other columns

pumpkins = pumpkins.drop([c for c in pumpkins.columns if c not in new_columns], axis = 1)
```

```python
# Calculate the average price from the high and low price columns

price = (pumpkins['Low Price'] + pumpkins['High Price']) / 2

# Get the month from the data column

month = pd.DatetimeIndex(pumpkins['Date']).month

# Create  a new data frame with these columns

new_pumpkins = pd.DataFrame({
    'Month': month,
    'Package': pumpkins['Package'], 
    'Low Price': pumpkins['Low Price'],
    'High Price': pumpkins['High Price'],
    'Price': price
})
```

```python
# Convert the price of all cells price by 1 1/9 bushels dividing by 1 1/9

new_pumpkins.loc[new_pumpkins['Package'].str.contains('1 1/9'), 'Price'] = price/(1 + 1/9)

#Convert the price of all cell price by 1/2 bushels divided by 1/2

new_pumpkins.loc[new_pumpkins['Package'].str.contains('1/2'), 'Price'] - price/(1/2)
```
